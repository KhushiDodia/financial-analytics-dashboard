```python
import pandas as pd
df = pd.read_csv('dirty_cafe_sales.csv')
df.head()
```

```python
df.info()
df.isnull().sum()
```

```python
df = df.replace(['ERROR', 'UNKNOWN'], pd.NA)
df.isnull().sum() 
```

```python
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'], errors='coerce')
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
df['Price Per Unit'] = pd.to_numeric(df['Price Per Unit'], errors='coerce')
df['Total Spent'] = pd.to_numeric(df['Total Spent'], errors='coerce')
df.dtypes
```

```python
print("Duplicates found:", df.duplicated().sum())
df = df.drop_duplicates()
```

```python
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
df['Price Per Unit'] = df['Price Per Unit'].fillna(df['Price Per Unit'].median())

# Total Spent can be recalculated instead of guessed
df['Total Spent'] = df['Quantity'] * df['Price Per Unit']

# Categorical columns: fill with the most common value (mode)
df['Payment Method'] = df['Payment Method'].fillna(df['Payment Method'].mode()[0])
df['Location'] = df['Location'].fillna(df['Location'].mode()[0])
df['Item'] = df['Item'].fillna(df['Item'].mode()[0])

df.isnull().sum()
```

```python
df.to_csv('cafe_sales_cleaned.csv', index=False)
print("Saved cleaned file. Final shape:", df.shape)
```

```python
df = pd.read_csv('cafe_sales_cleaned.csv') 
df['Transaction Date'] = pd.to_datetime(df['Transaction Date']) 
df.head()
```

```python
total_revenue = df['Total Spent'].sum()
total_transactions = len(df)
avg_transaction_value = df['Total Spent'].mean()
print(f"Total Revenue: ${total_revenue:,.2f}")
print(f"Total Transactions: {total_transactions}")
print(f"Average Transaction Value: ${avg_transaction_value:,.2f}")
```

```python
top_items = df.groupby('Item')['Total Spent'].sum().sort_values(ascending=False)
print(top_items)
```

```python
df['Month'] = df['Transaction Date'].dt.to_period('M')
monthly_revenue = df.groupby('Month')['Total Spent'].sum()
print(monthly_revenue)
```

```python
import matplotlib.pyplot as plt
monthly_revenue.plot(kind='line', marker='o', figsize=(10,5), title='Monthly Revenue Trend') 
plt.ylabel('Revenue ($)') 
plt.xlabel('Month') 
plt.show()
top_items.plot(kind='bar', figsize=(10,5), title='Revenue by Item') 
plt.ylabel('Revenue ($)') 
plt.show()
```
